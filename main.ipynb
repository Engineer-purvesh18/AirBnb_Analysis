# London AirBnB Analysis

## Data Loading Basic Exploration

```python
# Importing Required Libraries 

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

pd.set_option('display.max_columns', 50)
sns.set_style('whitegrid')
plt.rcParams['figure.figsize'] = (10, 5)

DATA_DIR = "AirBnb_Analysis"  # folder containing listings.csv, calendar.csv, reviews.csv
```

```python
listings = pd.read_csv(f"listings.csv", low_memory=False)
print("Shape:", listings.shape)
listings.head()
```

```python
listings.info()
```

```python
# Columns that are entirely (or almost entirely) empty in this export
fully_empty_cols = listings.columns[listings.isna().mean() == 1.0].tolist()
print(f"Dropping {len(fully_empty_cols)} fully-empty columns:")
print(fully_empty_cols)

# Columns that don't add analytical value (URLs, scrape metadata, free text, redundant quote fields)
not_useful_cols = [
    'listing_url', 'scrape_id', 'last_scraped', 'source', 'picture_url',
    'host_url', 'host_profile_url', 'host_profile_id', 'host_picture_url',
    'description', 'host_about', 'host_location', 'calendar_last_scraped',
    'price_quote_checkin_date', 'price_quote_checkout_date',
    'price_quote_total_price', 'price_quote_price_per_night', 'price_quote_raw',
]

cols_to_drop = [c for c in fully_empty_cols + not_useful_cols if c in listings.columns]
listings = listings.drop(columns=cols_to_drop)
print(f"\nDropped {len(cols_to_drop)} columns total. Remaining shape: {listings.shape}")
```

```python
# Missing value % for what's left
missing_pct = (listings.isna().mean() * 100).round(1).sort_values(ascending=False)
missing_pct[missing_pct > 0]
```

```python
listings['price'] = listings['price'].replace(r'[\$,]', '', regex=True).astype(float)

before = len(listings)
listings = listings.dropna(subset=['price'])
print(f"Dropped {before - len(listings)} rows with missing price ({(before-len(listings))/before:.1%})")
print(f"Remaining rows: {len(listings)}")

listings['price'].describe()
```

```python
for col in ['bedrooms', 'beds', 'bathrooms']:
    listings[col] = listings.groupby('room_type')[col].transform(lambda s: s.fillna(s.median()))
    # fallback: any group-level NaNs left (e.g. rare room type) -> overall median
    listings[col] = listings[col].fillna(listings[col].median())

listings[['bedrooms', 'beds', 'bathrooms']].isna().sum()
```

```python
listings['reviews_per_month'] = listings['reviews_per_month'].fillna(0)
for c in ['first_review', 'last_review']:
    listings[c] = pd.to_datetime(listings[c], errors='coerce')

print("Listings with zero reviews:", (listings['number_of_reviews'] == 0).sum(),
      f"({(listings['number_of_reviews']==0).mean():.1%} of all listings)")
```

### Data Cleaning And Basic Exploration

```python
listings['reviews_per_month'] = listings['reviews_per_month'].fillna(0)
for c in ['first_review', 'last_review']:
    listings[c] = pd.to_datetime(listings[c], errors='coerce')

print("Listings with zero reviews:", (listings['number_of_reviews'] == 0).sum(),
      f"({(listings['number_of_reviews']==0).mean():.1%} of all listings)")
```

```python
# Dropping Columns
# Small-missingness columns: drop the (few) affected rows rather than impute
small_missing_cols = [c for c in ['host_name', 'host_is_superhost', 'host_listings_count',
                                    'host_has_profile_pic', 'host_identity_verified',
                                    'bathrooms_text', 'has_availability'] if c in listings.columns]
before = len(listings)
listings = listings.dropna(subset=small_missing_cols)
print(f"Dropped {before - len(listings)} rows with sparse missing values")

# Duplicates
dupes = listings.duplicated(subset=['id']).sum()
listings = listings.drop_duplicates(subset=['id'])
print(f"Dropped {dupes} duplicate listing rows")

# Correct dtypes
bool_map = {'t': True, 'f': False}
for c in ['host_is_superhost', 'host_has_profile_pic', 'host_identity_verified', 'has_availability']:
    if listings[c].dtype == object:
        listings[c] = listings[c].map(bool_map)

listings['host_listings_count'] = listings['host_listings_count'].astype(int)

print("\nFinal cleaned shape:", listings.shape)
listings.dtypes.value_counts()
```

```python
listings.to_csv("listings_cleaned.csv", index=False)
listings.head()
```
