# London AirBnB Analysis

## Data Loading Basic Exploration

```python
# Importing Required Libraries 

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

pd.set_option('display.max_columns', 50)
sns.set_style('whitegrid')
plt.rcParams['figure.figsize'] = (10, 5)

DATA_DIR = "AirBnb_Analysis"  # folder containing listings.csv, calendar.csv, reviews.csv
```

```python
listings = pd.read_csv(f"listings.csv", low_memory=False)
print("Shape:", listings.shape)
listings.head()
```

```python
listings.info()
```

```python
# Columns that are entirely (or almost entirely) empty in this export
fully_empty_cols = listings.columns[listings.isna().mean() == 1.0].tolist()
print(f"Dropping {len(fully_empty_cols)} fully-empty columns:")
print(fully_empty_cols)

# Columns that don't add analytical value (URLs, scrape metadata, free text, redundant quote fields)
not_useful_cols = [
    'listing_url', 'scrape_id', 'last_scraped', 'source', 'picture_url',
    'host_url', 'host_profile_url', 'host_profile_id', 'host_picture_url',
    'description', 'host_about', 'host_location', 'calendar_last_scraped',
    'price_quote_checkin_date', 'price_quote_checkout_date',
    'price_quote_total_price', 'price_quote_price_per_night', 'price_quote_raw',
]

cols_to_drop = [c for c in fully_empty_cols + not_useful_cols if c in listings.columns]
listings = listings.drop(columns=cols_to_drop)
print(f"\nDropped {len(cols_to_drop)} columns total. Remaining shape: {listings.shape}")
```

```python
# Missing value % for what's left
missing_pct = (listings.isna().mean() * 100).round(1).sort_values(ascending=False)
missing_pct[missing_pct > 0]
```

```python
listings['price'] = listings['price'].replace(r'[\$,]', '', regex=True).astype(float)

before = len(listings)
listings = listings.dropna(subset=['price'])
print(f"Dropped {before - len(listings)} rows with missing price ({(before-len(listings))/before:.1%})")
print(f"Remaining rows: {len(listings)}")

listings['price'].describe()
```

```python
for col in ['bedrooms', 'beds', 'bathrooms']:
    listings[col] = listings.groupby('room_type')[col].transform(lambda s: s.fillna(s.median()))
    # fallback: any group-level NaNs left (e.g. rare room type) -> overall median
    listings[col] = listings[col].fillna(listings[col].median())

listings[['bedrooms', 'beds', 'bathrooms']].isna().sum()
```

```python
listings['reviews_per_month'] = listings['reviews_per_month'].fillna(0)
for c in ['first_review', 'last_review']:
    listings[c] = pd.to_datetime(listings[c], errors='coerce')

print("Listings with zero reviews:", (listings['number_of_reviews'] == 0).sum(),
      f"({(listings['number_of_reviews']==0).mean():.1%} of all listings)")
```

### Data Cleaning And Basic Exploration

```python
listings['reviews_per_month'] = listings['reviews_per_month'].fillna(0)
for c in ['first_review', 'last_review']:
    listings[c] = pd.to_datetime(listings[c], errors='coerce')

print("Listings with zero reviews:", (listings['number_of_reviews'] == 0).sum(),
      f"({(listings['number_of_reviews']==0).mean():.1%} of all listings)")
```

```python
# Dropping Columns
# Small-missingness columns: drop the (few) affected rows rather than impute
small_missing_cols = [c for c in ['host_name', 'host_is_superhost', 'host_listings_count',
                                    'host_has_profile_pic', 'host_identity_verified',
                                    'bathrooms_text', 'has_availability'] if c in listings.columns]
before = len(listings)
listings = listings.dropna(subset=small_missing_cols)
print(f"Dropped {before - len(listings)} rows with sparse missing values")

# Duplicates
dupes = listings.duplicated(subset=['id']).sum()
listings = listings.drop_duplicates(subset=['id'])
print(f"Dropped {dupes} duplicate listing rows")

# Correct dtypes
bool_map = {'t': True, 'f': False}
for c in ['host_is_superhost', 'host_has_profile_pic', 'host_identity_verified', 'has_availability']:
    if listings[c].dtype == object:
        listings[c] = listings[c].map(bool_map)

listings['host_listings_count'] = listings['host_listings_count'].astype(int)

print("\nFinal cleaned shape:", listings.shape)
listings.dtypes.value_counts()
```

```python
listings.to_csv("listings_cleaned.csv", index=False)
listings.head()
```

## Exploratory Data Analysis

```python
# Satistical Analysis

num_cols = ['price', 'accommodates', 'bedrooms', 'beds', 'bathrooms', 'minimum_nights',
            'number_of_reviews', 'reviews_per_month', 'review_scores_rating', 'availability_365']
listings[num_cols].describe().T
```

***Insight -***
Price Distribution is heavily right-skewd with some extreme outliers, so we look at both the raw and a 99th percentile clipped view

```python
fig, axes = plt.subplots(1, 2, figsize=(14, 5))
sns.histplot(listings['price'], bins=60, ax=axes[0], color='steelblue')
axes[0].set_title('Price distribution (raw)')

clip = listings['price'].quantile(0.99)
sns.histplot(listings.loc[listings['price'] <= clip, 'price'], bins=60, ax=axes[1], color='seagreen')
axes[1].set_title(f'Price distribution (clipped at 99th pct = £{clip:.0f})')
plt.tight_layout()
plt.show()
```

### Outlier Detection

```python
# Outlier check 
p95 = listings['price'].quantile(0.95)
n_outliers = (listings['price'] > p95).sum()
print(f"95th percentile price: £{p95:.2f} -- {n_outliers} listings ({n_outliers/len(listings):.1%}) priced above it")

plt.figure(figsize=(10, 5))
sns.boxplot(data=listings[listings['price'] <= listings['price'].quantile(0.99)],
            x='room_type', y='price', hue='room_type', palette='Set2', legend=False)
plt.title('Price by room type (99th pct clipped, for readability)')
plt.xticks(rotation=15)
plt.show()
```

### Listings per neighbourhood and room type

```python
top_neigh = listings['neighbourhood_cleansed'].value_counts().head(15)
plt.figure(figsize=(10, 6))
sns.barplot(x=top_neigh.values, y=top_neigh.index, hue=top_neigh.index, palette='viridis', legend=False)
plt.title('Top 15 neighbourhoods by number of listings')
plt.xlabel('Number of listings')
plt.show()
```

### Darta Distribution w.r.t room type

```python
plt.figure(figsize=(8, 5))
room_counts = listings['room_type'].value_counts()
plt.pie(room_counts, labels=room_counts.index, autopct='%1.1f%%', startangle=90,
        colors=sns.color_palette('pastel'))
plt.title('Share of listings by room type')
plt.show()
```

### Correlationn Heatmap

```python
corr = listings[num_cols].corr()
plt.figure(figsize=(9, 7))
sns.heatmap(corr, annot=True, fmt='.2f', cmap='coolwarm', center=0)
plt.title('Correlation matrix — numeric variables')
plt.tight_layout()
plt.show()
```

### Price vs Number of Views

```python
plt.figure(figsize=(9, 6))
sample_plot = listings[listings['price'] <= listings['price'].quantile(0.99)]
sns.scatterplot(data=sample_plot, x='number_of_reviews', y='price', hue='room_type',
                 alpha=0.4, s=20)
plt.title('Price vs. number of reviews')
plt.show()

print("Correlation (price, number_of_reviews):", listings['price'].corr(listings['number_of_reviews']).round(3))
```

### Room Type & Availability

```python
corr_avail_price = listings['availability_365'].corr(listings['price'])
print(f"Correlation between availability_365 and price: {corr_avail_price:.3f}")

plt.figure(figsize=(9, 6))
sns.scatterplot(data=sample_plot, x='availability_365', y='price', hue='room_type', alpha=0.4, s=20)
plt.title('Availability (days/year) vs. price')
plt.show()
```
